# Generate a report of all git repositories in a folder.

```python
import reporeviewer as rr

# 1. Fetch repos as plain Python dictionaries
myrepos = rr.getrepos('./repos/')
```

```python
from IPython.display import Markdown

# Render directly as rich notebook HTML/Markdown
print(f"There are {len(myrepos)} repos in the folder")
Markdown(rr.to_markdown(myrepos))
```

# NEEDS attention

```python
# 2. Filter using a comprehension
attention_repos = {
    name: data for name, data in myrepos.items() if rr.needs_attention(data)
}
print(f"{len(attention_repos)} REPOS NEED ATTENTION")
Markdown(rr.to_markdown(attention_repos))
```
